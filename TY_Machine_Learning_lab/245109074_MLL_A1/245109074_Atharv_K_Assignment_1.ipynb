```python
def average(values):
    total=0
    for v in values:
        total+=v
    return total/len(values)

print(average(test1))
print(average(test2))
print(average(attendance))
```

### B4. A First Taste of Geometry

#### **Task:** Solution

```python
meera=np.array([88,90,95])
kabir=np.array([62,58,70])
dist=np.sqrt(np.sum((meera-kabir)**2))
print(dist)
```

```python
nisha_index=names.index("Nisha")
nisha=np.array([test1[nisha_index],test2[nisha_index],attendance[nisha_index]])
min_dist=float("inf")
closest=""
for i in range(len(names)):
    if names[i]=="Nisha":
        continue
    student=np.array([test1[i],test2[i],attendance[i]])
    d=np.sqrt(np.sum((nisha-student)**2))
    if d<min_dist:
        min_dist=d
        closest=names[i]
print(closest,min_dist)
```

### D1. Grading Function

```python
def grade(marks):
    if marks>=85:
        return "A"
    elif marks>=70:
        return "B"
    elif marks>=50:
        return "C"
    else:
        return "F"

df["grade"]=df["test2"].apply(grade)
print(df["grade"].value_counts())
```

### D3. Variance From Scratch

```python
mean_t1=sum(test1)/len(test1)
total=0
for v in test1:
    total+=(v-mean_t1)**2
variance=total/len(test1)
print(variance)
print(np.var(t1))
print(round(variance,6)==round(np.var(t1),6))
```

### D2. A Mistyped Entry

```python
df.loc[df["name"]=="Rohan","test1"]=400
print(df["test1"].mean())
print(df["test1"].std())
```

### **Answer:** 
so changing Rohan's test1 value to 400 makes the mean sharply upward since it's now far larger than every other score.

the standard deviation increases very high, since one value is much big then others.

this shows how a single outlier or data-entry error can change summary statistics for an entire dataset.
